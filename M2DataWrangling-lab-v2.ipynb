```python
# Write your code here
print(df.dtypes)
print(df.info())
print(df.isnull().sum())
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
print(df['Country'].unique())

# Count occurrences of each country
print(df['Country'].value_counts())

# Identify missing or inconsistent entries
inconsistent_country = df[
    (df['Country'].isnull()) | 
    (df['Country'].str.strip() == '') |
    (df['Country'].str.lower().isin(['unknown', 'n/a', 'na']))
]

print(inconsistent_country)
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
country_mapping = {
    'USA': 'United States',
    'U.S.A': 'United States',
    'United States of America': 'United States',
    'Viet Nam': 'Vietnam'
}
df['Country'] = df['Country'].replace(country_mapping)
print(df['Country'].value_counts())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
# One-hot encode the Employment column
employment_encoded = pd.get_dummies(df['Employment'], prefix='Employment')
# Join encoded columns back to the original dataframe
df = pd.concat([df, employment_encoded], axis=1)
# Optional: remove the original Employment column
df = df.drop('Employment', axis=1)
# Display updated dataframe
print(df.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
# Count missing values in each column
missing_values = df.isnull().sum()

# Sort columns by highest number of missing values
missing_values_sorted = missing_values.sort_values(ascending=False)

#Display the result
print(missing_values_sorted)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# Impute missing values in ConvertedCompYearly with median
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(
    df['ConvertedCompYearly'].median()
)
print(df['ConvertedCompYearly'].isnull().sum())
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# Impute missing values in RemoteWork with the most frequent value (mode)
most_frequent = df['RemoteWork'].mode()[0]

df['RemoteWork'] = df['RemoteWork'].fillna(most_frequent)

#V Verify missing values are removed
print(df['RemoteWork'].isnull().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
!pip install scikit-learn
from sklearn.preprocessing import MinMaxScaler

# Create scaler object
scaler = MinMaxScaler()

# Apply Min-Max Scaling
df['ConvertedCompYearly_Normalized'] = scaler.fit_transform(
    df[['ConvertedCompYearly']]
)

# Display results
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_Normalized']].head())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

# Apply Log transformation
df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])

# Display results
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_Log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
# Create ExperienceLevel column based on YearsCodePro

def categorize_experience(years):
    try:
        years = float(years)

        if years < 3:
            return 'Beginner'
        elif years < 7:
            return 'Intermediate'
        else:
            return 'Experienced'
            
    except:
        return 'Unknown'

# Apply function to create new column
df['ExperienceLevel'] = df['YearsCodePro'].apply(categorize_experience)

# Display results
print(df[['YearsCodePro', 'ExperienceLevel']].head())
```
